# BookHaven — Recommender System Exploration

This notebook walks through data loading, feature engineering, and model training for the hybrid recommender.

```python
import sys, os
sys.path.insert(0, os.path.abspath('../..'))
from dotenv import load_dotenv
load_dotenv('../../.env')

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
sns.set_theme(style='whitegrid')
```

```python
from data.db_connect import get_books_df, get_orders_df, get_sessions_df, get_reviews_df

books    = get_books_df()
orders   = get_orders_df()
sessions = get_sessions_df()
reviews  = get_reviews_df()

print(f'Books: {len(books)}  Orders: {len(orders)}  Sessions: {len(sessions)}  Reviews: {len(reviews)}')
```

```python
from ml.data.interactions import build_interactions
interactions = build_interactions()
print(interactions.describe())
interactions.head(10)
```

```python
from ml.recommender.content import build_content_model, get_similar_books
content_model = build_content_model(books)
first_book_id = books.iloc[0]['id']
print(f'Similar to "{books.iloc[0]["title"]}":')
similar = get_similar_books(first_book_id, content_model, top_n=5)
for bid in similar:
    title = books[books['id']==bid]['title'].values
    print(' -', title[0] if len(title) else bid)
```

```python
from ml.recommender.collab import build_collab_model
collab_model = build_collab_model(interactions)
print('RMSE:', collab_model.get('rmse'))
print('Trained:', collab_model.get('trained'))
```

```python
from ml.recommender.train import run_training
metrics = run_training()
print('Training complete:', metrics)
```
